```python
#Nhóm thư viện cơ bản
import numpy as np              #Xử lý số học, mảng và ma trận
import pandas as pd             #làm việc với dữ liệu dạng bảng(CSV, Excel)
import matplotlib.pyplot as plt #Vẽ biểu đồ cơ bản
#try:
import seaborn as sns           # Vẽ biểu đồ đẹp và trực quan hơn
# except ModuleNotFoundError:
#     sns = None

#Nhóm thư viện sklearn (Machine learning)
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,             #độ chính xác
    classification_report,      #báo cáo precision, recall, f1-score
    roc_auc_score,              #chỉ số ROC AUC
    roc_curve,                  #vẽ đường cong ROC
    precision_recall_curve,     #vẽ đường cong Precision_recall
)
```

```python
from tensorflow.keras.datasets import mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Tạo và lưu 10 ảnh (từ 0 đến 9)
for i in range(10):
    # Tìm bức ảnh đầu tiên tương ứng với chữ số i
    idx = np.where(y_train == i)[0][0]
    img = X_train[idx]
    
    # Lưu ảnh ra thư mục dự án dưới dạng đen trắng
    file_name = f'image_{i}.png'
    plt.imsave(file_name, img, cmap='gray')

print("✅ Đã xuất thành công 10 ảnh từ image_0.png đến image_9.png!")

print("Kích thước bộ dữ liệu:")
print(f"Kích thước tập train: {X_train.shape}, {y_train.shape}")
print(f"Kích thước tập test: {X_test.shape}, {y_test.shape}")
```

```python
from collections import Counter #colections là thư viện chứa các cấu trúc dữ liệu đặc biệt, Counter là một trong số đó, dùng để đếm tần suất xuất hiện của các phần tử trong một iterable (danh sách, tuple, chuỗi,...)

#đém tần suất hiển thị của tững nhãn
counter = Counter(y_train)
counter = counter.most_common()         #most_common() trả về danh sách các phần tử và số lần xuất hiện của chúng, sắp xếp theo thứ tự giảm dần
counter = pd.DataFrame(counter)         #đưa vào DataFrame để dễ dàng hiển thị
counter.columns = ['Số', 'Số lượng']    #đặt tên cột

display(counter) #hiển thị DataFrame
```

```python
from tensorflow.keras.utils import to_categorical

#====== Preprocess cho CNN ======
def preprocess_for_cnn(X_train, X_test, y_train, y_test):
    #reshape dữ liệu đầu vào cho CNN
    X_train = X_train.reshape(-1, 28, 28, 1) #thêm chiều thứ 4 cho ảnh xám
    X_test = X_test.reshape(-1, 28, 28, 1)

    #chuẩn hóa dữ liệu đầu vào
    X_train = X_train.astype('float32') / 255.0
    X_test = X_test.astype('float32') / 255.0

    #one-hot encoding cho nhãn
    y_train = to_categorical(y_train, 10)
    y_test = to_categorical(y_test, 10)

    return X_train, X_test, y_train, y_test
```

```python
X_train, X_test, y_train, y_test = preprocess_for_cnn(X_train, X_test, y_train, y_test)
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import(
    Input, Conv2D, MaxPooling2D,
    Flatten, Dense, Dropout
)

#built model
model = Sequential([
    Input(shape=(28,28,1)),

    #convolution Block 1
    Conv2D(64, kernel_size=(3,3), activation='relu'),
    MaxPooling2D(pool_size=(2,2)),

    #convolution Block 2
    Conv2D(128, kernel_size=(3,3), activation='relu'),
    MaxPooling2D(pool_size=(2,2)),

    #flatten và dense layers
    Flatten(),

    Dense(64, activation='relu'),
    Dropout(0.5),

    Dense(32, activation='relu'),
    Dropout(0.3),

    #output layer (10 classes)
    Dense(10, activation='softmax')
])

#compile model
model.compile( #complie là lệnh để biên dịch mô hình, xác định hàm mất mát, bộ tối ưu hóa và các chỉ số đánh giá
    optimizer='adam', #adam tối ưu hóa bằng cách
    loss='categorical_crossentropy', #categorical_crossentropy là hàm mất mát được sử dụng cho bài toán phân loại đa lớp
    metrics=['accuracy'] #accuracy là chỉ số đánh giá để đo lường hiệu suất của mô hình
)

#Summary of the model
model.summary()
```

```python
model.fit( #fit dùng để huấn luyện mô hình trên dữ liệu đầu vào, trong đó X_train là dữ liệu đầu vào, y_train là nhãn tương ứng, epochs là số lần lặp qua toàn bộ dữ liệu, batch_size là số lượng mẫu được sử dụng để cập nhật trọng số của mô hình trong mỗi lần lặp, validation_split là tỷ lệ dữ liệu huấn luyện được sử dụng để đánh giá mô hình trong quá trình huấn luyện, verbose là mức độ hiển thị thông tin huấn luyện
    X_train, y_train,
    epochs= 10,
    batch_size= 32, #nghĩa là số lượng mẫu được sử dụng để cập nhật trọng số của mô hình trong mỗi lần lặp. Trong trường hợp này, batch_size=32 có nghĩa là mô hình sẽ sử dụng 32 mẫu dữ liệu để tính toán gradient và cập nhật trọng số một lần. Việc sử dụng batch_size nhỏ giúp mô hình học nhanh hơn và giảm thiểu việc bị mắc kẹt trong các cực tiểu cục bộ, nhưng cũng có thể làm tăng độ nhiễu trong quá trình học. Ngược lại, batch_size lớn giúp mô hình học ổn định hơn nhưng có thể làm chậm quá trình học.
    validation_split= 0.2, # tham số này có ý nghĩa là 20% dữ liệu huấn luyện sẽ được sử dụng để đánh giá mô hình trong quá trình huấn luyện, giúp theo dõi hiệu suất của mô hình trên dữ liệu chưa thấy trước đó và tránh overfitting.
    verbose=1 #nếu =0 nghĩa là không hiển thị thông tin huấn luyện, =1 hiển thị thanh tiến trình, =2 hiển thị thông tin chi tiết về mỗi epoch
)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

def pre_and_eva(model, X_test, y_test):
    y_pred = model.predict(X_test) #dự đoán nhãn cho tập test

    #nếu OP là xs(CNN) -> convert sang label
    if len(y_pred.shape) > 1:  #nếu y_pred có nhiều hơn 1 chiều (tức là đầu ra của mô hình là one-hot encoding), thì ta cần chuyển đổi nó sang nhãn thực tế bằng cách lấy chỉ số của giá trị lớn nhất trong mỗi vector one-hot.
        y_pred = np.argmax(y_pred, axis=1) # convert từ one-hot encoding sang label

    #nếu y_test là one-hot encoding -> convert sang label
    if len(y_test.shape) > 1:  #nếu y_test có nhiều hơn 1 chiều (tức là đầu ra của mô hình là one-hot encoding), thì ta cần chuyển đổi nó sang nhãn thực tế bằng cách lấy chỉ số của giá trị lớn nhất trong mỗi vector one-hot.
        y_test = np.argmax(y_test, axis=1) # convert từ one-hot encoding sang label

    acc = accuracy_score(y_test, y_pred)*100 #tính độ chính xác bằng cách so sánh nhãn dự đoán với nhãn thực tế, nhân với 100 để chuyển đổi sang phần trăm
    cfs = confusion_matrix(y_test, y_pred) #tạo ma trận nhầm lẫn
    clr = classification_report(y_test, y_pred) #tạo báo cáo phân loại

    print(f"Độ chính xác: {acc:.4f}%")
    print("-"*50)
    print("Ma trận nhầm lẫn:")
    print(cfs)
    print("-"*50)
    print("Báo cáo phân loại:")
    print(clr)
```

```python
#Gọi hàm để đánh giá mô hình
pre_and_eva(model, X_test, y_test)
```

```python
# @title Văn bản tiêu đề mặc định
from PIL import Image, ImageOps #PIL là thư viện xử lý ảnh, Image là lớp để làm việc với ảnh, ImageOps là lớp để thực hiện các thao tác xử lý ảnh như đảo màu, cắt, xoay, thay đổi kích thước,...

import numpy as np

def input_image(img_array): #hàm được dùng để xử lý ảnh đầu vào từ người dùng, nhận vào một mảng numpy (img_array) và trả về một ảnh PIL đã được xử lý sẵn để dự đoán.
    if img_array is None:
        return None

    if isinstance(img_array, dict): #nghĩa là nếu img_array là một dict (từ điển) thì ta sẽ lấy giá trị của key 'composite' nếu có, nếu không có thì lấy giá trị của key 'image'. Nếu img_array không phải là dict thì giữ nguyên giá trị img_array.
        img_array = img_array.get('composite', img_array.get('image'))

    

    img = Image.fromarray(img_array.astype(np.uint8)) #để chuyển đổi mảng numpy thành ảnh PIL, ta sử dụng Image.fromarray() và ép kiểu dữ liệu về uint8 để đảm bảo rằng các giá trị pixel nằm trong khoảng từ 0 đến 255.

    if img.mode == "RGBA":
        bg = Image.new("RGB", img.size, (255, 255, 255))
        bg.paste(img, mask=img.split()[3])
        img = bg.convert("L")
    elif img.mode == "RGB":
        img = img.convert("L")

    if np.mean(img) > 128:
        img = ImageOps.invert(img)

    image_preview = img.resize((28, 28), Image.LANCZOS)
    return image_preview


def preprocess_image(img):
    if img is None:
        return None
    return preprocess_input(img)


def preprocess_input(img):
    """
    img: image_preview
    """
    img = np.array(img).astype('float32') / 255.0
    img = img.reshape(1, 28, 28, 1)
    return img

```

```python
import matplotlib.pyplot as plt
import numpy as np

def predict_character(img_input, model_choice='CNN'): #CNN đã được huấn luyện
    # Sử dụng global để truy cập biến model từ các cell khác
    global model

    # Kiểm tra xem biến model đã được định nghĩa chưa
    if 'model' not in globals():
        return "❌ Lỗi: Bạn chưa chạy cell huấn luyện mô hình! Hãy chạy cell định nghĩa 'model' trước.", None, ""

    image_preview = input_image(img_input)
    if image_preview is None:
        return 'Vui lòng vẽ hoặc tải ảnh!', None, ''

    img = preprocess_image(image_preview)
    if img is None:
        return 'Vui lòng vẽ hoặc tải ảnh!', None, ''

    try:
        # Lấy model từ biến toàn cục
        model_selected = model
        probs = model_selected.predict(img, verbose=0)[0]

        top5_idx = probs.argsort()[::-1][:5] #lấy 5 chỉ số có xác suất cao nhất, argsort() sắp xếp các chỉ số theo thứ tự tăng dần, [::-1] đảo ngược để có thứ tự giảm dần, [:5] lấy 5 phần tử đầu tiên
        top5_label = [str(i) for i in top5_idx] #chuyển đổi chỉ số sang chuỗi
        top5_prob = [float(probs[i]) for i in top5_idx] #lấy xác suất tương ứng với 5 chỉ số cao nhất

        best_label = top5_label[0] #lấy nhãn có xác suất cao nhất
        best_conf = top5_prob[0] #lấy xác suất cao nhất

        fig, ax = plt.subplots(figsize=(6, 3))
        bars = ax.barh(top5_label[::-1], top5_prob[::-1])
        ax.set_xlim(0, 1)
        ax.set_xlabel('Confidence')
        ax.set_title('Top-5 Predictions')

        for bar, prob in zip(bars, top5_prob[::-1]):
            ax.text(
                bar.get_width() + 0.01,
                bar.get_y() + bar.get_height() / 2,
                f'{prob:.2%}',
                va='center'
            )

        plt.tight_layout()

        top5_text = '\n'.join( 
            [f'{i + 1}. {lbl}: {p:.2%}' for i, (lbl, p) in enumerate(zip(top5_label, top5_prob))]
        ) #tạo chuỗi mô tả 5 dự đoán tốt nhất
        result_text = f'🎯 Dự đoán: **{best_label}** ({best_conf:.2%})'

        return result_text, fig, top5_text
    except Exception as e:
        return f"❌ Lỗi khi dự đoán: {str(e)}", None, ""
```
