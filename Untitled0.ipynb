```python
from tensorflow.keras.datasets import mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()

print("Kích thước bộ dữ liệu:")
print(f"Kích thước tập train: {X_train.shape}, {y_train.shape}")
print(f"Kích thước tập train: {X_test.shape}, {y_test.shape}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(10, 10, figsize=(10, 20))  #tạo nhiều ảnh con trên 1 khung hình cụ thể là 10x10, tỷ lệ ảnh 10x20

for digit in range (10):
    idx = np.where(y_train == digit)[0][:10] #cú pháp np.where(y_train == digit) trả về các chỉ số của các phần tử trong mảng y_train có giá trị bằng digit. [0] lấy mảng đầu tiên (chỉ số), [:10] lấy 10 phần tử đầu tiên
    
    for i, ax in enumerate(axes[digit]):
        ax.imshow(X_train[idx[i]].reshape(28, 28), cmap='gray') #hiển thị ảnh xám
        ax.set_title(f"{digit}") #đặt tiêu đề cho ảnh con
        ax.axis('off') #tắt trục tọa độ
        
plt.tight_layout() #tự động điều chỉnh khoảng cách giữa các ảnh con
plt.show() #hiển thị khung hình
```

```python
from collections import Counter #colections là thư viện chứa các cấu trúc dữ liệu đặc biệt, Counter là một trong số đó, dùng để đếm tần suất xuất hiện của các phần tử trong một iterable (danh sách, tuple, chuỗi,...)

#đém tần suất hiển thị của tững nhãn
counter = Counter(y_train)
counter = counter.most_common()         #most_common() trả về danh sách các phần tử và số lần xuất hiện của chúng, sắp xếp theo thứ tự giảm dần
counter = pd.DataFrame(counter)         #đưa vào DataFrame để dễ dàng hiển thị
counter.columns = ['Số', 'Số lượng']    #đặt tên cột

display(counter) #hiển thị DataFrame
```

```python
#sắp xếp theo cột số
counter_sorted = counter.sort_values(by='Số')

#vẽ biểu đồ cột
plt.figure()
plt.bar(counter_sorted['Số'], counter_sorted['Số lượng'])

#nhãn và tiêu đêf
plt.xlabel('Chữ số')

#ép hiển thị đẩy đủ các chữ số trên trục x
plt.xticks(range(10))
plt.ylabel('Số lượng ảnh')
plt.title('Phân phối nhãn trong MNIST')

#hiển thị
plt.show()
```

```python
from tensorflow.keras.utils import to_categorical

#====== Preprocess cho CNN ======
def preprocess_for_cnn(X_train, X_test, y_train, y_test):
    #reshape dữ liệu đầu vào cho CNN
    X_train = X_train.reshape(-1, 28, 28, 1) #thêm chiều thứ 4 cho ảnh xám
    X_test = X_test.reshape(-1, 28, 28, 1)
    
    #chuẩn hóa dữ liệu đầu vào
    X_train = X_train.astype('float32') / 255.0
    X_test = X_test.astype('float32') / 255.0
    
    #one-hot encoding cho nhãn
    y_train = to_categorical(y_train, num_classes=10) 
    y_test = to_categorical(y_test, num_classes=10) 
    
    return X_train, X_test, y_train, y_test
```
